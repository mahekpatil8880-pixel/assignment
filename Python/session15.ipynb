```python
#task-1
def get_song_duration_per_minute(total_duration,number_of_songs):
    try:
        duration_per_song=total_duration / number_of_songs
        return duration_per_song
    except ZeroDivisionError:
        return "Number of song cannot be zero."

print(get_song_duration_per_minute(300,0))
print(get_song_duration_per_minute(200,6))
```

```python
#task-2
def cal(total_cart_amount,item_count):
    try:
        price_per_item=total_cart_amount / item_count
        return price_per_item
    except ZeroDivisionError:
        return "Item count cannot be zero"

print(cal(2000,20))
print(cal(2999,0))
```

```python
#task-3
def cal(total_spend,number_of_offer):
    try:
        if number_of_offer==0:
            raise NoOffersApplied("No offers have been applied.")
        
        avg_cashback=total_spend / number_of_offer
        return avg_cashback
        
    except NoOffersApplied as e:
        return str(e)

print(cal(2030,10))
print(cal(2999,0))
```

```python
#task-4
def calculate_average_rating(total_rating, num_reviews):
    try:
        cal=total_rating / num_reviews
        return cal
    except ZeroDivisionError:
        return "Error: Number of reviews cannot be zero."
    finally:
        print("Thank you for using the calculator")

print(calculate_average_rating(500, 0))
```

```python
#task-5
def safe_divide_for_zomato(bill_amount,number_of_people):
    try:
        bill=bill_amount / number_of_people
    except ZeroDivisionError:
        return "Number of people cannot be zero."
    else:
        return f"Bill per person:{bill}"
    finally:
        print("Split calculation done")

print(safe_divide_for_zomato(2000,0)) 
print(safe_divide_for_zomato(1900,4))       
```
