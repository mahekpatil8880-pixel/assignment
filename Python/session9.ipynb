```python
#task-1
price=12000
discount_rate=10
def calculate_final_price(price, discount_rate):
    discount=price*discount_rate/100
    final_price=price-discount
    return final_price

calculate_final_price(price, discount_rate)
```

```python
#task-3
def format_price(price, currency='INR'):
    if currency=='INR':
        return '₹500'
    elif currency=='USD':
        return '$500'

print(format_price(3000))
print(format_price(2000,'USD'))
print(format_price(2000,'INR'))
```

```python
#task-4
discount_rate=10
def apply_coupon(price, coupon_code=None):
    if coupon_code=='ZOMATO10':
        discount=price*discount_rate/100
        final_price=price-discount
        return final_price
    else:
        return price

print(apply_coupon(1000))
print(apply_coupon(1000,'ZOMATO10'))
```
