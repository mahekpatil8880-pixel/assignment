```python
#task-3
import csv

f1=open("ipl_score.csv","w",newline="")


ipl=csv.writer(f1)
ipl.writerow(["match_id","team1","team2","winner"])
ipl.writerow([1,"CSK","MI","CSK"])
ipl.writerow([2,"RCB","KKR","KKR"])
ipl.writerow([3,"GT","RR","GT"])
ipl.writerow([4,"MI","DC","MI"])
ipl.writerow([5,"CSK","RCB","RCB"])

f1.close()
print("CSV file created")
```

```python
import csv

f1=open("ipl_score.csv","r")
ipl=csv.DictReader(f1)

for row in ipl:
    print("winner:",row["winner"])

f1.close()
```

```python
#task-4
import json

f1=open("movie.json","w")
data=[
    {"title":"3 Idiots","year":"2010","rating":8},
    {"title":"Dangal","year":"2015","rating":9},
    {"title":"tare zameen par","year":"2013","rating":9.5}
]

json.dump(data,f1)
f1.close()
print("json file created")
```

```python
import json

f1=open("movie.json","r")
data=json.load(f1)

for movie in data:
    print("Title:",movie["title"])
    print("Rating:",movie["rating"])
f1.close()
```

```python
#task-5
from pathlib import Path
import json

f1=Path("my_fav_apps.json")

if not f1.exists():
    data=[
        {"name":"Instagram","category":"Social Media"},
        {"name":"Zomato","category":"Food Delivery"},
        {"name":"Paytm","category":"Finance"}
    ]

    f2=open(f1,"w")
    json.dump(data,f2)
    f2.close()

    print("File created")
else:
    print("File already exists")
```
