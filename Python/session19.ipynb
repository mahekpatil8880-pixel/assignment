```python
#task-1
class Song:
    def __init__(self,title,artist,duration):
        self.title=title
        self.artist=artist
        self.duration=duration

s=Song("Darkhasst","Arijit singh","180")

print("Title:",s.title)
print("Artist:",s.artist)
print("Duration:",s.duration,"sec")
```

```python
#task-3
class Foodorder:
    def __init__(self,restaurant_name,items,total_price):
        self.restaurant_name=restaurant_name
        self.items=items
        self.total_price=total_price

food=Foodorder("tulsi",["paneer tikka","chapati"],400)

print("restaurant_name:",food.restaurant_name)
print("items:",food.items)
print("total_price:",food.total_price)
```

```python
#task-4
class Foodorder:
    def __init__(self,restaurant_name,items,total_price):
        self.restaurant_name=restaurant_name
        self.items=items
        self.total_price=total_price

    def add_item(self, item_name, item_price):
        self.item_name=item_name
        self.item_price=item_price
        self.items.append(item_name)
        self.total_price+=item_price

food=Foodorder("tulsi",["paneer tikka","chapati"],400)

food.add_item("lassi",100)
food.add_item("gulab jamun",120)

print("restaurant_name:",food.restaurant_name)
print("items:",food.items)
print("total_price:",food.total_price)
```
