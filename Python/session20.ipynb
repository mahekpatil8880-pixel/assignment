```python
#task-1
class Playlist:
    def __init__(self):
        self._songs=[]

    def add_song(self,song):
        self._songs.append(song)

p=Playlist()

p.add_song("Darkhasst")
p.add_song("Kesariya")
p.add_song("Tum Hi Ho")

print("Playlist:", p._songs)
```

```python
#task-2
class Product:
    def __init__(self,name,price):
        self.name=name
        self._price=price

    def get_price(self):
        return self._price

    def set_price(self,price):    
        self._price=price

p=Product("mobile","20000")
print("old price:",p.get_price())

p.set_price(23000)
print("updated price:",p.get_price())
```

```python
#task-3
class Movie:
    def __init__(self,title,rating):
        self.title=title
        self._rating=rating

    def getter(self):
        return self._rating

    def setter(self,rating):
        if 0<=rating and rating<=10:
            self._rating=rating
        else:
            print("Error: Rating must be between 0 and 10.")

m=Movie("3 Idiots",7)
print("rating:",m.getter())

m.setter(8)
print("updated rating:",m.getter())

m.setter(23)
```

```python
#task-4
from abc import ABC, abstractmethod

class Paymentmethod:
    @abstractmethod
    def pay(self,amount):
        pass

class Paytm(Paymentmethod):
    def pay(self,amount):
        print(f"paid {amount} using paytm")

class Phonepe(Paymentmethod):
    def pay(self,amount):
        print(f"paid {amount} using phonepe")

paytm=Paytm()
phonepe=Phonepe()

paytm.pay(200)
phonepe.pay(3000)
```
