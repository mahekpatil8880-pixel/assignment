```python
#task-2
class Product:
    def __init__(self,name,price):
        self.name=name
        self.price=price

    def get_discounted_price(self):
        dis_price=(self.price*10)/100
        return self.price-dis_price
        
class Electronics(Product):
    def get_discounted_price(self):
        dis_price=(self.price*20)/100 
        return self.price-dis_price

product=Electronics("mobile",10000)
print("Original Price:", product.price)
print("Price:",product.get_discounted_price())
```

```python
#task-3
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_discounted_price(self):
        dis_price = (self.price * 10) / 100
        return self.price - dis_price


class Electronics(Product):
    def get_discounted_price(self):
        dis_price = (self.price * 20) / 100
        return self.price - dis_price


def show_final_price(item):
    print("Name:", item.name)
    print("Discounted Price:", item.get_discounted_price())


# Create objects
product = Product("Shirt", 2000)
electronics = Electronics("Mobile", 10000)

# Demonstrate polymorphism
show_final_price(product)
show_final_price(electronics)
```

```python
#task-4
class Ticket:
    def __init__(self, movie_name, price):
        self.movie_name = movie_name
        self.price = price

    def get_final_price(self):
        return self.price


class PremiumTicket(Ticket):
    def get_final_price(self):
        return super().get_final_price() + 50


# Create objects
ticket = Ticket("3 Idiots", 200)
premium_ticket = PremiumTicket("3 Idiots", 300)

# Print final prices
print("Normal Ticket:", ticket.get_final_price())
print("Premium Ticket:", premium_ticket.get_final_price())
```
