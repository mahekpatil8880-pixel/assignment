```python
import json

songs = [
    {"song": "Blinding Lights", "artist": "The Weeknd", "streams": 3500000},
    {"song": "Shape of You", "artist": "Ed Sheeran", "streams": 4200000},
    {"song": "Levitating", "artist": "Dua Lipa", "streams": 2800000}
]

with open("trending_songs.json", "w") as file:
    json.dump(songs, file, indent=4)

print("JSON file created successfully!")
```

```python
import pandas as pd

data = {
    "Product_ID": range(1, 10001),
    "Product_Name": [f"Product_{i}" for i in range(1, 10001)],
    "Price": [100 + (i % 5000) for i in range(1, 10001)]
}

df = pd.DataFrame(data)

df.to_excel("flipkart_products.xlsx", index=False)

print("Excel file created successfully!")
```

```python
df = pd.read_excel("flipkart_products.xlsx")

chunk_size = 2000

for start in range(0, len(df), chunk_size):
    chunk = df.iloc[start:start + chunk_size]
    print("Rows in chunk:", len(chunk))
```

```python
with open("paytm_transactions.csv", "w") as file:
    file.write("TransactionID;Amount;Category;Status\n")
    file.write("P101;500;Food;Success\n")
    file.write("P102;;Recharge;Success\n")
    file.write("P103;299;Shopping;\n")
    file.write("P104;1000;Bills;Success\n")

print("CSV file created successfully!")
```
