```python
#task2
df = pd.read_csv("ipl_players.csv")

print("Shape before dropping:", df.shape)

df_clean = df.dropna(axis=0, how='any')

print("Shape after dropping:", df_clean.shape)

print(df_clean)
```

```python
#task3
df = pd.read_csv("ipl_players.csv")

df["runs"] = df["runs"].fillna(df["runs"].mean())

print(df["runs"])
```

```python
#task4
df = pd.DataFrame({
    "restaurant": ["Pizza Hub", "Burger Point", "Food Villa", "Spice House", "Cafe Time"],
    "rating": [4.5, None, None, 4.2, None]
})

print("Before Filling:")
print(df)

df["rating"] = df["rating"].ffill()


df["rating"] = df["rating"].bfill()

print("\nAfter Filling:")
print(df)
```
