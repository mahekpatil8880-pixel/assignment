```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
np.random.seed(42)
delivery_times = np.random.randint(20, 61, 50)

sns.histplot(delivery_times, bins=8, kde=True)


plt.title('Distribution of Zomato Delivery Times')
plt.xlabel('Delivery Time (minutes)')
plt.ylabel('Number of Orders')

plt.show()
```

```python
tips = sns.load_dataset('tips')

print(tips.head())


plt.figure(figsize=(8, 5))

sns.boxplot(
    x='day',
    y='total_bill',
    data=tips
)


plt.title('Distribution of Total Bill by Day')
plt.xlabel('Day of the Week')
plt.ylabel('Total Bill Amount ($)')

plt.show()
```

```python
sns.set_theme(style="darkgrid")
teams = ['MI', 'CSK', 'RCB', 'KKR', 'GT', 'RR', 'SRH', 'DC']


np.random.seed(42)

data = []

for team in teams:
    scores = np.random.normal(loc=170, scale=25, size=20)
    
    for score in scores:
        data.append([team, round(score)])
df = pd.DataFrame(data, columns=['Team', 'Runs'])

print(df.head())
plt.figure(figsize=(10, 6))

sns.violinplot(
    x='Team',
    y='Runs',
    data=df
)

plt.title('IPL Match Run Distribution by Team')
plt.xlabel('IPL Team')
plt.ylabel('Runs')

plt.show()
```
