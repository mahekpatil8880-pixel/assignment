```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']
sales = [20, 35, 30, 45, 50]
products = ['A', 'B', 'C', 'D']
orders = [25, 40, 30, 50]
hours = [1, 2, 3, 4, 5, 6]
marks = [40, 45, 55, 60, 70, 80]
categories = ['Food', 'Travel', 'Shopping', 'Other']
expenses = [40, 25, 20, 15]
fig, axes = plt.subplots(2, 2, figsize=(10, 8))


axes[0, 0].plot(days, sales, marker='o')
axes[0, 0].set_title('Daily Sales')
axes[0, 0].set_xlabel('Day')
axes[0, 0].set_ylabel('Sales')


axes[0, 1].bar(products, orders)
axes[0, 1].set_title('Product Orders')
axes[0, 1].set_xlabel('Product')
axes[0, 1].set_ylabel('Orders')

axes[1, 0].scatter(hours, marks)
axes[1, 0].set_title('Study Hours vs Marks')
axes[1, 0].set_xlabel('Study Hours')
axes[1, 0].set_ylabel('Marks')

axes[1, 1].pie(expenses, labels=categories, autopct='%1.1f%%')
axes[1, 1].set_title('Monthly Expenses')


plt.tight_layout()
plt.show()
```

```python
apps = ['Zomato', 'Swiggy', "Domino's"]
delivery_time = [32, 28, 35]


colors = ['red', 'orange', 'blue']


plt.figure(figsize=(8, 5))

bars = plt.bar(apps, delivery_time, color=colors,
               edgecolor='black', linewidth=2)

plt.title('Average Delivery Time Comparison')
plt.xlabel('Food Delivery App')
plt.ylabel('Average Delivery Time (minutes)')

for bar in bars:
    plt.text(bar.get_x() + bar.get_width()/2,
             bar.get_height(),
             str(bar.get_height()),
             ha='center', va='bottom')

plt.show()
```

```python
influencers = ['Aarav', 'Priya', 'Rahul', 'Neha', 'Karan']

followers = [12000, 25000, 18000, 40000, 32000]
daily_posts = [2, 4, 3, 5, 4]

fig, ax1 = plt.subplots(figsize=(9, 5))


ax1.plot(influencers, followers,
         color='blue', marker='o', linewidth=2,
         label='Followers')

ax1.set_xlabel('Instagram Influencers')
ax1.set_ylabel('Number of Followers', color='blue')
ax1.tick_params(axis='y', labelcolor='blue')


ax2 = ax1.twinx()

ax2.plot(influencers, daily_posts,
         color='red', marker='s', linewidth=2,
         label='Average Daily Posts')

ax2.set_ylabel('Average Daily Posts', color='red')
ax2.tick_params(axis='y', labelcolor='red')

plt.title('Instagram Followers vs Average Daily Posts')
plt.show()
```

```python

df = pd.DataFrame({
    'Movie': ['Dhurandhar', 'Dhamaal 4', 'Border 2', 'Cocktail 2', 'Mardaani 3'],
    'Tickets_Sold': [85, 70, 75, 55, 45],
    'IMDb_Rating': [8.2, 6.9, 5.8, 6.5, 6.0]
})

print(df)

plt.figure(figsize=(9, 5))

plt.scatter(df['Tickets_Sold'], df['IMDb_Rating'],
            color='purple', marker='o', s=100)

for i in range(len(df)):
    plt.annotate(
        df['Movie'][i],
        (df['Tickets_Sold'][i], df['IMDb_Rating'][i]),
        xytext=(0, 8),
        textcoords='offset points',
        ha='center'
    )

plt.title('Movie Tickets Sold vs IMDb Ratings')
plt.xlabel('Tickets Sold (in lakhs)')
plt.ylabel('IMDb Rating')
plt.grid(True)

plt.show()
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

sales = [120, 135, 150, 145, 160, 175,
         180, 190, 210, 230, 400, 250]

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(months, sales, marker='o')
max_sales = max(sales)
max_index = sales.index(max_sales)

ax.annotate(
    'Big Billion Days',
    xy=(months[max_index], max_sales),
    xytext=(months[max_index], max_sales + 50),
    arrowprops=dict(arrowstyle='->'),
    fontsize=12
)

ax.set_title("Flipkart Monthly Sales")
ax.set_xlabel("Month")
ax.set_ylabel("Sales (₹ Crores)")

plt.grid(True)
plt.show()
```
