```python
import pandas as pd 
```

```python
#task1
orders = {
    'restaurant_name': [
        'Pizza Hut',
        'Dominos',
        'Burger King',
        'Pizza Hut',
        'McDonalds',
        'Dominos',
        'Subway'
    ],
    'order_date': [
        '2026-09-20',
        '2026-09-21',
        '2026-09-22',
        '2026-09-20',  
        '2026-09-23',
        '2026-09-21',  
        '2026-09-24'
    ]
}
df = pd.DataFrame(orders)


df.to_csv('zomato_orders.csv', index=False)

print("Zomato Order History:")
print(df)

duplicates = df[df.duplicated(
    subset=['restaurant_name', 'order_date'],
    keep=False
)]

print("\nDuplicate Orders:")
print(duplicates)
```

```python
#task4
df = pd.DataFrame({
    'username': [
        'insta_queen',
        'insta-queen',
        'instaqueen',
        'photo_lover',
        'insta_queen',
        'travel_guru',
        'insta-queen'
    ]
})

print("Original DataFrame:")
print(df)

df['username'] = df['username'].replace({
    'insta_queen': 'instaqueen',
    'insta-queen': 'instaqueen'
})

print("\nDataFrame After Standardization:")
print(df)
```

```python
#task5
df = pd.DataFrame({
    'payment_status': [
        'Yes', ' yes ', 'Y', 'No', ' no ', 'N',
        'YES', ' y ', 'NO', ' n '
    ]
})

print("Original DataFrame:")
print(df)


df['payment_status'] = (
    df['payment_status']
    .str.strip()      
    .str.lower()      
    .replace({
        'yes': 1,
        'y': 1,
        'no': 0,
        'n': 0
    })
)


print("\nCleaned DataFrame:")
print(df)
```
