```python
import pandas as pd
import matplotlib as plt
import matplotlib
print(matplotlib.__version__)
import matplotlib
print(matplotlib.__file__)
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({
    'order_amount': [250, 320, 180, 450, 290, 350, 400, 275, 310, 500, 1200, 1500]
})

print("Swiggy Orders Dataset:")
print(df)

plt.figure(figsize=(6, 4))
plt.boxplot(df['order_amount'])


plt.title('Boxplot of Swiggy Order Amounts')
plt.xlabel('Order Amount')
plt.ylabel('Amount (₹)')

plt.show()

print("Outliers are the points that appear outside the whiskers of the boxplot.")
print("In this dataset, 1200 and 1500 are visible outliers.")
```
