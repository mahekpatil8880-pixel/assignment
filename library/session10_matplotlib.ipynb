```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
days = ['Day 1', 'Day 2', 'Day 3', 'Day 4', 'Day 5', 'Day 6', 'Day 7']
steps = [4500, 6200, 5800, 7100, 8900, 7600, 9400]


plt.figure(figsize=(8, 4))
plt.plot(days, steps, marker='o')


plt.title('Daily Steps Over Last 7 Days')
plt.xlabel('Day')
plt.ylabel('Number of Steps')


plt.savefig('steps_lineplot.png')

plt.show()
```

```python
df = pd.DataFrame({
    'Restaurant': [
        'Spice Villa', 'Royal Dine', 'Food Hub', 'Pizza Point',
        'Tasty Treat', 'Green Bowl', 'Urban Kitchen',
        'Cafe Aroma', 'BBQ Nation', 'Street Eats'
    ],
    'Zomato_Rating': [4.5, 4.2, 3.8, 4.0, 4.7, 3.9, 4.3, 4.1, 4.6, 3.7],
    'Average_Meal_Price': [800, 650, 400, 550, 900, 450, 700, 600, 1000, 350]
})
print(df)


plt.figure(figsize=(8, 5))
plt.scatter(df['Zomato_Rating'], df['Average_Meal_Price'])


for i in range(len(df)):
    plt.text(
        df['Zomato_Rating'][i],
        df['Average_Meal_Price'][i],
        df['Restaurant'][i],
        fontsize=8
    )
plt.xlabel('Zomato Rating')
plt.ylabel('Average Meal Price (₹)')
plt.title('Restaurant Ratings vs Average Meal Price')


plt.show()
```

```python
apps = ['Swiggy', 'Zomato', "Domino's"]

orders = [18, 25, 12]
colors = ['orange', 'red', 'blue']
plt.figure(figsize=(7, 5))

for i in range(len(apps)):
    plt.bar(apps[i], orders[i], color=colors[i], label=apps[i])

plt.title('Food Delivery Orders in the Last Month')
plt.xlabel('Food Delivery App')
plt.ylabel('Number of Orders')
plt.legend()
plt.show()
```

```python
durations = [
    25, 40, 35, 60, 45,
    30, 55, 70, 20, 50,
    65, 35, 80, 45, 40,
    90, 30, 55, 75, 50
]


plt.figure(figsize=(8, 5))
plt.hist(durations, bins=5, color='purple', edgecolor='black')


plt.title('Spotify Listening Session Durations')
plt.xlabel('Duration (minutes)')
plt.ylabel('Number of Sessions')
plt.show()
```

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']


screen_time = [2.5, 3.0, 2.0, 3.5, 4.0, 5.0, 4.5]  # hours
posts_liked = [20, 25, 15, 30, 35, 45, 40]

fig, axes = plt.subplots(2, 1, figsize=(8, 8))

axes[0].plot(days, screen_time, marker='o')
axes[0].set_title('Daily Instagram Screen Time')
axes[0].set_xlabel('Day')
axes[0].set_ylabel('Screen Time (hours)')
axes[1].bar(days, posts_liked)
axes[1].set_title('Instagram Posts Liked Each Day')
axes[1].set_xlabel('Day')
axes[1].set_ylabel('Number of Posts Liked')

plt.tight_layout()
plt.show()
```
